# Simple scripts to count the number of stable voxels in a dataset

```python
import os

import nibabel as nib
import numpy as np

data_path = "/Volumes/T7/BIDS-BRAINPLAYBACK-TASK2/derivatives/mvpa_08_stability_bold/stab_full_masks/"

n_subjects = 20

# initalize numpy array to hold counts
counts = np.zeros(n_subjects)

for ss in range(1, n_subjects + 1):
    # Construct the file path for the mask

    mask_path = os.path.join(data_path, f"sub-{ss:02d}_STAB_mask.nii.gz")

    # Load the mask using nibabel
    mask_img = nib.load(mask_path)

    # count the number of non-zero voxels
    counts[ss - 1] = (mask_img.get_fdata() != 0).sum()

counts
```

```python
np.mean(counts), np.std(counts), np.min(counts), np.max(counts)
# Output the results
print(f"Mean: {np.mean(counts)}")
print(f"Standard Deviation: {np.std(counts)}")
print(f"Minimum: {np.min(counts)}")
print(f"Maximum: {np.max(counts)}")
```
